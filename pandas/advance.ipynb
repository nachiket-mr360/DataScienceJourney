```python
import pandas as pd
import numpy as np
data = {
    'name':['alice','bob','charlie','David','Emma'],
    'math': [88,75,92,60,85],
    'science':[90,80,78,70,95]
}
df = pd.DataFrame(data)
df
```

```python
#apply() function, we can you function for all data using apply
df['total marks'] = df[['math','science']].apply(lambda row: row.sum(), axis=1)
df
```

```python
col_mean = df[['math','science']].apply(lambda col : col.mean())
col_mean
```

```python
def to_grade(score):
    if score>=90:
        return 'grade A'
    elif score >=80:
        return 'grade B'
    elif score >=35:
        return 'grade C'
    else :
        return 'fail'
    
df['maths grade'] = df['math'].apply(to_grade)
df
```

```python
df['science grade'] = df['science'].apply(to_grade)
```

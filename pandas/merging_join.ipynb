```python
import pandas as pd
```

```python
'''
1.using concat() for stacking data
* pd.concate() combine dataframes vertically or horizontally
use axis=0 for vertical stacking default, axis=1 for horizontal.
ignore_index = true can reset the index after concatenation'''
df1 = pd.DataFrame({
    "ProductID" : [101,102,103],
    "product name": ["pen", "notebook",'marker'],
    'price':[10,50,30]
})
df2 = pd.DataFrame({
    'ProductID' : [104,105],
    'product name':['pencil','eraser'],
    'price':[5,8]
})
df3 = pd.DataFrame({
    'stock':[100,60,80],
    'rating':[4.5,4.7,4.8]
})
```

```python
print(df1)
print(df2)
print(df3)
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
pd.concat([df1,df3]) #get nan when we have unsimilar data
```

```python
pd.concat([df1,df3], axis=1) #concate horizantally, hence not get nan
```

```python
''' 
using merge() for SQL-style joins
* merge() is similar to sql joins
* specify on or left_on/right_on to choose key columns
* join types: inner default, left, right, outer'''
products = pd.concat([df1,df2], ignore_index=True)
products
```

```python
sales = pd.DataFrame({
    'productID': [101,102,104,106], #106 doesnt exist in products
    'UnitsSold':[500,200,150,50]
})
```

```python
pd.merge(products, sales, left_on='ProductID', right_on='productID')
```

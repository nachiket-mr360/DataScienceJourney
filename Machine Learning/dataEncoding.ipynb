```python
#nominal/one hot encoding > it breate binary vector for each categories
#example :single, married, seperated
#single : [1,0,0]
#married : [0,1,0]
#separated : [0,0,1]
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({'status':['single','married','separated','single','single','separated','separated','married']})
encoder = OneHotEncoder()
encoder
```

```python
encoded = encoder.fit_transform(df[['status']]).toarray()
encoded
```

```python
encoder.get_feature_names_out()
```

```python
encoder_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
encoder_df
```

```python
#we can remove one columns, we when we know the two remaining columns the 3rd column automaticall get it's values
#so in one hot encoding always n-1 columns we can keep, one column always drop
```

```python
df
```

```python
encoder_df
```

```python
df = pd.concat([df,encoder_df],axis=1)
```

```python
import seaborn as sns
df = sns.load_dataset('tips')
df.head()
```

```python
#convert sex and smoker into onehot encoding.
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
result= encoder.fit_transform(df[['sex','smoker']]).toarray()
encoder_df2 = pd.DataFrame(result, columns=['is_female','is_male','not_smoker','is_smoker'])
df = pd.concat([df,encoder_df2],axis=1)
df
```

```python
#target guided ordinal encoding
#based on relationship with target variable, and we have lot of unique categories
#when we are unable to apply one hot encoding
#replace the cate with the mean and median of respective group target
```

```python
df=pd.DataFrame({'time':['lunch','bf','dinner','dinner','dinner','bf','lunch','dinner','dinner','bf','lunch','dinner','dinner','bf','lunch',],
              'total_bill':[120,130,120,80,90,100,125,85,90,100,120,110,100,90,85]})
df
```

```python
mean_price = df.groupby('time')['total_bill'].mean().to_dict()
mean_price
```

```python
df['time_encoded'] = df['time'].map(mean_price)
df
```

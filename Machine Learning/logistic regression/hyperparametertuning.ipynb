```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
X,y = make_classification(n_samples=1000, n_features=10, n_redundant=5, n_informative=5, n_classes=2, random_state=1)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
```

```python
#hyperparameter tunning
#there are two types in it.
#1.Grid search CV
from sklearn.model_selection import GridSearchCV
```

```python
params= {'penalty':('l1','l2','elasticnet'), 'C':[1,2,10,20,30,40]}
classifier = LogisticRegression()
classifier
```

```python
clf = GridSearchCV(classifier, param_grid=params,cv=5)
clf
```

```python
import warnings
warnings.filterwarnings('ignore')
clf.fit(x_train,y_train)
```

```python
clf.best_params_
```

```python
clf.best_score_
```

```python
model = LogisticRegression(C=1,penalty='l2' )
model
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
model.predict_proba(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
#randamized search cv, it will take randomly some combinations
from sklearn.model_selection import RandomizedSearchCV
randomModel = RandomizedSearchCV(classifier,param_distributions=params, cv=5, n_iter=10)
randomModel
```

```python
randomModel.fit(x_train, y_train)
```

```python
randomModel.best_params_
```

```python
randomModel.best_score_
```

```python
model = LogisticRegression(C=10, penalty='l2')
model
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

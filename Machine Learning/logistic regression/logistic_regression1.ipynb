```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df
```

```python
df['target'] = data.target
```

```python
df.head()
```

```python
df.tail()
```

```python
df.sample()
```

```python
df.target.unique()
```

```python
#for binary classification drop class3
df_final = df[df['target'] !=2]
df_final['target'].unique()
```

```python
df_final
```

```python
#take x and y
X = df_final.iloc[:,:-1]
y = df_final.iloc[::,-1]
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train , x_test, y_train, y_test= train_test_split(X,y , test_size=0.2,random_state=1)
x_train.shape, x_test.shape
```

```python
#scalling (optional not required here)
#model training
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression()
classifier
```

```python
classifier.fit(x_train, y_train)
```

```python
y_pred = classifier.predict(x_test) #it will get class not probability
y_pred
```

```python
#to get probability
classifier.predict_proba(x_test) #corresponding to higher probability class is predicted cutoff is 0.5
```

```python
#evaluation matrix
#confusion matrics
from sklearn.metrics import confusion_matrix, accuracy_score,  classification_report
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

```python
#first two rows (0 and 1) >> precision, recall,f1 score and no of data point of respectie class
#macro avg -> gives overall model's precision/recall/f1-score 
#weighted avg-> used in case of class imbalance, it gives importance performance on dominant class
```

```python
from sklearn.metrics import roc_curve, auc
probability = classifier.predict_proba(x_test)[:,1] #probability of class 1 (we may use 0 also)
probability
```

```python
#ROC (reviever operating characteristics curve) it shows relation between 
# tPR and FPR with different cutoff of probabilityes
#input y_test,y_predicted probability 
fpr, tpr, thresholds = roc_curve(y_test,probability)
#how well your model isable to distibuse between two classes
```

```python
fpr,tpr, thresholds
```

```python
#calculate AUC curve
roc_auc =auc(fpr,tpr)
roc_auc
```

```python
plt.plot(fpr,tpr, color='darkorange', linewidth=2, label='roc curve (area=%0.2f)'%roc_auc)
plt.plot([0,1],[0,1],color='navy', linewidth=2, linestyle='--')
plt.xlim([0.0,1.0])
plt.ylim([0.0,1.05])
```

```python
#cross validation : two waysto do 1.logisticregressionCV 2.use kfold
from sklearn.linear_model import LogisticRegressionCV
```

```python
from sklearn.model_selection import KFold
cv = KFold(n_splits=5,shuffle=True)
cv
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(classifier,x_train,y_train, cv=cv, scoring='accuracy')
scores
```

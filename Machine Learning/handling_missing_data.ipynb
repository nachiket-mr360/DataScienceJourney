How to deal with missing values/missing value treatment (should be asked from business team)
1.	If the data is huge and missing value is less than 1% of data, you can drop the missing the value.
2.	Impute/replace the missing values (if we don’t want to drop the data)
if the data is continuous in nature then we can impure it with mean or median 
(if outlier treatment is not done then impute the missing value with median), 
if the data is categorical in nature than with can impute it with mode 
3.	In any column if the missing value is greater than 40%, drop the column.
(why to drop, if we impute the data it’s not nature assurance it our own assumption which might be wrong)
4.	Impute missing value with some random numbers/0/number which is not possible
when we are imputing (handling missing values) we are introducing noise in the data) so it prefer to only handle less amount of data to get more accurate results.

1)missing completely at random (MCAR):The data are missing is independent of observed and unobserved data (ex: there is a survey, if a person is suffering from a disease or not, let’s we have lots of participant here 
Here the person 3 and 5 don’t know whether they have disease or not that’s the data is missing there, those missing values are random they both are independent (not dependent to any of data ) (they don’t have any relation with each other or with any other in data) those types of missing data is called as MCAR (missing completely at random).here the missing data is randomly distributed throughout the data.

2)missing at random (MAR):The missing data depends on the observed data but not on the missing data itself. (ex. Survey to know the income of people, lets we check the family income, and one of the family member didn’t want to disclose the income due to age/gender or another factor but not due to income level.)

 3)missing data not at random (MNAR):
missing values depends on the value of missing data itself. (ex: if we are collecting the data of income and job satisfaction of employees in a company, 1.the employees who are not satisfied in the company will refuse to report their income, then the data is not missing at random)


```python
#age column is missing at random because it relationship other column (survived) (they died hence we didnot got age)
#trying to impute it with mean, but before imputing with mean, check outlier presence
sns.boxenplot(df['age'])
```

```python
sns.histplot(df.age, kde=True)
```

```python
#as we see there is to negligible outliers, and the curve is looks like a normal distribution
#when we are getting normal distribution always impute the data with mean
```

```python
df['mean_imputation'] = df['age'].fillna(df['age'].mean())
```

```python
df[['mean_imputation','age']]
```

```python
sns.histplot(df['mean_imputation'],kde=True)
```

```python
#mean imputation works well with normal distribution or no outlier. 
#if outlier present then we go for median imputation
df['median_imputation'] = df['age'].fillna(df['age'].median())
```

```python
df[['median_imputation','mean_imputation','age']]
```

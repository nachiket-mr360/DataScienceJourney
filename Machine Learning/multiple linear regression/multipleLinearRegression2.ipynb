```python
X = df.iloc[:,:-1] #independent variables
y =df.iloc[:,-1] #target/dependent variable
X
```

```python
y
```

```python
#train test split
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y ,test_size=0.2, random_state=1)
x_train.shape, x_test.shape, y_train.shape , y_test.shape
```

```python
#scalling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#model training
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model
```

```python
model.fit(x_train, y_train)
```

```python
x_train.shape
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred= model.predict(x_test)
```

```python
#evaluation
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mse= mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test, y_pred)
score= r2_score(y_test,y_pred)
rmse = np.sqrt(mean_squared_error(y_test,y_pred))
print(mse,mae, score,rmse)
```

```python
#adjusted R score 1-(1-rscore)/(n-p-1)
adjR=1-(1-score)*(len(y_test)-1)/(len(y_test)-x_test.shape[1]-1)
adjR
```

python pickle module is used for serialising and de-serialising a python object structure, any object in python can be pickled so that it can be saved on disk, what pckle does is that it "serialises" the object first before writing it to file, pickling is a way to convert a python object (list, dict,etc ) into a character stream. the idea is that this character stream contains allthe information necessary to reconstruct the object in another python script.

```python
model
```

#pickle >> allows you to store the python object on the disk, it's a convinient way to store and transform complex python object
#serialization>> it's a process of converting a data structure r anobjectinto a format that canbe easily stored, transmitted or persisted, during the serialization, the object's state is converted intoa stream of bytes or other formats that can be written to disk , sent over a network or stored in 
#de serialization>> deserialization is theprocess of reconstructing a serialized object back into its original form, during deserialization the serialized data is readand convertedbackinto a data structure or object that matches theoriginal

```python
import pickle
pickle.dump(model,open("model.pkl","wb")) #write binary mode
#python object(model) it's attributes and methods is coverted into byte steam
```

```python
model2 = pickle.load(open('model.pkl','rb'))
model2
```

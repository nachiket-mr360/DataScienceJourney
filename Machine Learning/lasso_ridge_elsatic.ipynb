```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('mpg')
df
```

```python
df.drop('name',axis=1, inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df['horsepower'].fillna(df['horsepower'].median(),inplace=True)
df.isnull().sum()
```

```python
df.info()
```

```python
#there is one object and all numerical columns. we need to change the object into numeric

df.dtypes
```

```python
df['origin'].value_counts()
```

```python
df['origin']= df['origin'].map({'usa':1, "japan":2,'europe':3})
df['origin'] = df['origin'].astype(int)
df.dtypes
```

```python
#separate x and y
X = df.drop('mpg',axis=1)
y = df['mpg']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=1)
x_train ,x_test,y_train, y_test
```

```python
#simple linear regression model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model
```

```python
model.fit(x_train,y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f'the coefficient for {col_name} is {model.coef_[i]}')
```

```python
#coeffients arerelativly smaller,means if one independent variable changes slightly 
#there willl be not much difference in prediction
#this is sametime is called smoother model
from sklearn.metrics import r2_score
y_pred = model.predict(x_test)
score = r2_score(y_test, y_pred)
score
```

```python
# ridge regression
from sklearn.linear_model import Ridge
ridgemodel =Ridge(alpha=0.1)
ridgemodel
```

```python
ridgemodel.fit(x_train,y_train)
for i,col_name in enumerate(x_train.columns):
    print(f'the coefficient for {col_name} is {ridgemodel.coef_[i]}')
```

```python
y_pred_ridge = ridgemodel.predict(x_test)
r2_ridge = r2_score(y_test,y_pred_ridge)
r2_ridge
```

```python
#we don't see much variation in coeff of ridge regression as compared to linear regression
```

```python
from sklearn.linear_model import Lasso
lasso_model = Lasso(alpha=0.5)
lasso_model.fit(x_train,y_train)
```

```python
for i,col_name in enumerate(x_train.columns):
    print(f'the coefficient for {col_name} is {lasso_model.coef_[i]}')
```

```python
#3 featurescoefficient are 0, because lasso helps in feature selection (so we may drop them for better model)
y_pred_lasso = lasso_model.predict(x_test)
r2_lasso = r2_score(y_test,y_pred_lasso)
r2_lasso
```

```python
#elastic net : use both advantage of ridge and lasso
from sklearn.linear_model import ElasticNet
```

```python
elasticmodel = ElasticNet(alpha=1,l1_ratio=0.5)
```

```python
elasticmodel.fit(x_train,y_train)
for i,col_name in enumerate(x_train.columns):
    print(f'the coefficient for {col_name} is {elasticmodel.coef_[i]}')
```

```python
y_pred_elastic = elasticmodel.predict(x_test)
r2_elastic = r2_score(y_test,y_pred_elastic)
r2_elastic
```

```python
from sklearn.linear_model import RidgeCV
ridgecvModel = RidgeCV(cv=5)
ridgecvModel.fit(x_train,y_train)
y_pred_rcv = ridgecvModel.predict(x_test)
r2_rcv = r2_score(y_test,y_pred_rcv)
r2_rcv
```

```python
ridgecvModel.get_params()
```

```python
from sklearn.linear_model import LassoCV
LassoCVModel = LassoCV(cv=5)
LassoCVModel.fit(x_train,y_train)
y_pred_rcv = LassoCVModel.predict(x_test)
r2_Lcv = r2_score(y_test,y_pred_rcv)
r2_Lcv
```

```python
from sklearn.linear_model import ElasticNetCV
elasticCVModel = ElasticNetCV(cv=5)
elasticCVModel.fit(x_train,y_train)
y_pred_ecv = elasticCVModel.predict(x_test)
r2_ecv = r2_score(y_test,y_pred_rcv)
r2_ecv
```

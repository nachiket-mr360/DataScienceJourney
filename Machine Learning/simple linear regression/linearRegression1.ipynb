```python
#let's construct the dataframe
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
```

```python
df
```

```python
df['target'] = diabetes.target
```

```python
df
```

```python
#EDA, data cleaning,data preparing, feature engineering (but here the data is already performed all those thing so we move to ml directly)
```

```python
#divide data into X,Y (for simple linear regression only one feature)
x = df[['bmi']]
y = df[['target']]
```

```python
#split the data into train-test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=1)
#random_state used to get same data points everytime
#this will return 4 things (x_train, x_test, y_train, y_test)
```

```python
#here we need to do scalling but data is already in the scale (optional)
```

```python
#model training
from sklearn.linear_model import LinearRegression
```

```python
#how to build a model
#step1 : make an object of model/instaniate an LR model
model = LinearRegression()
model
```

```python
#step 2:train the model
model.fit(x_train, y_train) #at backend it will learn mathematical relationship,it will use gredient descent to give optimal coefients 
#it will learn miu and sigma and then we perform trainform 
#.fit will train the model (try to understand mathematical relationship (in case of linear regression
# Ordinary list square -> MSE then gradient descent then we get thita0, thita1)
#.predict using thita0, thita1 make prediction
#linear regression always accept data in the form of dataframes
```

```python
model.coef_ #give coeffient B1
```

```python
model.intercept_ #give intercept B0
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
#visualize the results
plt.scatter(x_test,y_test, color='black',label='actual data')
plt.plot(x_test, y_pred, color='blue',linewidth=3, label='linear regression line')
plt.xlabel('bmi')
plt.ylabel('one year progression target')
plt.title('linear regression ob biabetes data')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_test,y_pred)
```

```python
#in z-test we make a null hypothesis, if that P value less then 0.05 then we reject it
null_mean = 0.09 #assume that we have mean 0.09
#p-value <0.05 >> reject null hypothesis else we fail to reject null hypothesis.
#and if Z-stats > Z-critical then we reject it else we fail to reject this.
```

```python
#for Z-test those are things we must know
'''statsmodels.stats.weightstats.ztest¶
statsmodels.stats.weightstats.ztest(x1, x2=None, value=0, alternative='two-sided', usevar='pooled', ddof=1.0)[source]¶
test for mean based on normal distribution, one or two samples

In the case of two samples, the samples are assumed to be independent.

Parameters¶
x1 : array_like, 1-D or 2-D¶
first of the two independent samples

x2 : array_like, 1-D or 2-D¶
second of the two independent samples

value : float¶
In the one sample case, value is the mean of x1 under the Null hypothesis. In the two sample case, value is the difference between mean of x1 and mean of x2 under the Null hypothesis. The test statistic is x1_mean - x2_mean - value.

alternative : str¶
The alternative hypothesis, H1, has to be one of the following

’two-sided’: H1: difference in means not equal to value (default) ‘larger’ : H1: difference in means larger than value ‘smaller’ : H1: difference in means smaller than value

usevar : str, 'pooled' or 'unequal'¶
If pooled, then the standard deviation of the samples is assumed to be the same. If unequal, then the standard deviation of the sample is assumed to be different.

ddof : int¶
Degrees of freedom use in the calculation of the variance of the mean estimate. In the case of comparing means this is one, however it can be adjusted for testing other statistics (proportion, correlation)

Returns¶
tstat (float) – test statistic

pvalue (float) – pvalue of the t-test

Notes

usevar can be pooled or unequal in two sample case'''
```

```python
from statsmodels.stats.weightstats import ztest
#statsmodel is library which have all the tests
```

```python
Zscore, pvalue = ztest(population, value=null_mean, alternative='larger', ) #alternative say's that mean is greater then null mean
Zscore
```

```python
pvalue
```

```python
if pvalue <= 0.05:
    print('reject the null hypothesis')
else:
    print('fail to reject null hypothesis')
```

```python
'''question
in a school calculated IQ scores of 50 students, the average IQ turned out to be 100, 
the mean of population IQ is 90 and std is 16, state whether the clain by school
that IQ increases if student study more then average school, with 5% of
significant level'''
'''ans:
given n=50, mean = 90, x=100, std = 16, alpha=0.05
theoratically 
Z-statc = x-mean/std/root(n)
        = 100-90 / 16/root(50)
null_hypothesis : mean_iq = 90
alternate : mean_iq > 90'''

```

```python
#practically using Z-score and Zcritical
import scipy.stats as stats #scipy is library where we have stats module containing all tests

sample_mean = 100
population_mean = 90
population_std = 16
sample_size = 50
alpha = 0.05


```

```python
Zscore = (sample_mean - population_mean)/ (population_std/np.sqrt(sample_size))
print(f'the Zscore is : {Zscore}')
```

```python
#find Z-critial (to find Z-critial we use the same library scipy.stats)
stats.norm.ppf 
'''stats >> it's stats module in scipy library, norm refers to normal distribution
ppf >> parcent point function -- it's nothing but inverse of cummulative distribution funtion(cdf)
'''
Zcritical = stats.norm.ppf(1-alpha) #giving Z-critical
Zcritical
```

```python
if Zscore > Zcritical:
    print('reject the null hypothesis')
else: 
    print('fail to reject null hypothesis')
```

```python
pvalue = 1-stats.norm.cdf(Zscore)
```

```python
if pvalue < alpha:
    print('reject the null hypothesis')
else:
    print('fail to reject null hypothesis')
```

```python
#one sample T-test
population = np.random.randint(10,50, 50)
population
```

```python
np.mean(sample)
```

```python
null_value = 23 #assuming average mean
import scipy

Tscore, Pvalue = scipy.stats.ttest_1samp(sample, null_value) #bydefault it will 95% confidence (alpha=0.05)
print(Tscore, Pvalue)
```

```python
#if pvalue > 0.05 fail to reject null hypothesis.
if Pvalue>0.05:
    print('fail to reject null hypothesis')
else:
    print('reject null hypothesis')
```

```python
#chi-squared test (it's for categorical data)
data = sns.load_dataset('tips')
data
```

```python
data[['sex','smoker']]
```

```python
data_table = pd.crosstab(data['sex'], data['smoker']) #it will give cummulative table
data_table
```

```python
observed_values = data_table.values #to see only values
observed_values
```

```python
import scipy.stats as stats #using chi2 critical and stats
stats.chi2_contingency(observed_values)
#gives 4 things, stats_test, pvalue, dof, expected values
stats_test, Pvalue, dof, expected_value = stats.chi2_contingency(observed_values)
print(stats_test, Pvalue, dof, expected_value)
```

```python
#in chi2 test, we have two things observed and expected values
expected_value
```

```python
observed_values
```

```python
#dof = (number of rows -1 ) * (number of col -1)
dof = 1 #because 2 row and 2 col
```

```python
from scipy.stats import chi2
```

```python
for i in zip(observed_values, expected_value): #combine both 
    print(i) #each of the row gives observed and expected values 
```

```python
chisqaure_test = sum([(o-e)**2/e for o,e in zip(observed_values, expected_value)])
```

```python
chisqaure_test #get two values, so summetation of it is chisquare
```

```python
chisqaure_stats = chisqaure_test[0] + chisqaure_test[1]
chisqaure_stats
```

```python
dof 
```

```python
alpha
```

```python
chi2_critical = chi2.ppf(1-alpha, df=dof)
chi2_critical
```

```python
if chisqaure_stats >= chi2_critical:
    print('reject the null hypothesis, there is a relationship between two categorical variable')
else:
    print('fail to reject null hypothesis, no relationship')
```

```python
#using pvalue method
Pvalue = 1-chi2.cdf(x=chisqaure_stats, df=dof)
Pvalue
```

```python
#if pvalue > 0.05 fail to reject null hypothesis.
if Pvalue>0.05:
    print('fail to reject null hypothesis')
else:
    print('reject null hypothesis')
```

```python
#find the critical value
alpha = 0.05
dof = len(observed_values)-1
critical_Value = stats.chi2.ppf(1-alpha, dof)
critical_Value
```

```python
if chisqaure_stats > critical_Value:
    print('reject the null hypothesis')
else:
    print('fail to reject null hypothesis')
```

```python
#F-test 
'''there are two workers, if one worker is effective more then other one.'''
worker1 = [18,17,20,25,27,55,51,42,46,29]
worker2 = [18,16,22,26,39,51,48,44,58,12]
```

```python
#to calculate F-stats
Fstats = np.var(worker1)/np.var(worker2)
Fstats
```

```python
dof1 = len(worker1)-1
dof2 = len(worker2) -1
alpha = 0.05
dof1, dof2, alpha
```

```python
import scipy.stats as stat
critical_value =stat.f.ppf(q=1-alpha, dfn=dof1,dfd=dof2)
critical_value
```

```python
if Fstats > critical_value:
    print('reject the null hypothesis')
else:
    print('fail to reject null hypothesis')
```

```python
'''Scenario

A company introduced a new training program for its employees. They want to know whether the training improved employee productivity.

They randomly selected two independent groups:

Group A → Employees who did not receive the training.
Group B → Employees who received the training.

The productivity scores are:

group_A = [72, 75, 70, 68, 74, 71, 69, 73]

group_B = [78, 80, 79, 77, 81, 82, 76, 80]

Use:

alpha = 0.05'''
```

```python
import numpy as np
from scipy.stats import ttest_ind

group_A = [72, 75, 70, 68, 74, 71, 69, 73]

group_B = [78, 80, 79, 77, 81, 82, 76, 80]
alpha = 0.05
Ho = "mean group A = mean group B"
Ha = "Mean group A is not equal to mean group B."
Tstats, Pvalue = ttest_ind(group_A, group_B)
print(f'T-stats ={Tstats} and P-value={Pvalue}')
if Pvalue > alpha:
    print(f'fail to reject null hypothesis {Ho}')
else:
    print(f'reject the null hypothesis {Ha}')
    
print(globals())
```

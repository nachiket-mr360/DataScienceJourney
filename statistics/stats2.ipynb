```python
import pandas as pd
import numpy as np
import seaborn as sns
import statistics as st
```

```python
#to define random population
population = np.random.randint(10,20,50)
#np.random.randint(start, end, datapoint)
population
```

```python
len(population)
```

```python
#checking mean mode, median
print(f"the mean of the population {st.mean(population)}")
print(f"the mode of population is :{st.mode(population)}")
print(f"the median of the population is : {st.median(population)}")
```

```python
#generating random sample from population 
sample = np.random.choice(population, 20)
#np.random.choice(data, datapoint_required)
sample
```

```python
#checking mean, mode and median for both sample and population 
print(f"the mean of the population {st.mean(population)}")
print(f"the mode of population is :{st.mode(population)}")
print(f"the median of the population is : {st.median(population)}")
print(f"the mean of the sample {st.mean(sample)}")
print(f"the mode of sample is :{st.mode(sample)}")
print(f"the median of the sample is : {st.median(sample)}")


```

```python
#taking different samples
sample1 = np.random.choice(population, 20)
sample2 = np.random.choice(population, 20)
sample3 = np.random.choice(population, 20)
sample4 = np.random.choice(population, 20)
```

```python
#checking all sample means
mean_of_samples = []
all_samples = [sample1, sample2, sample3, sample4]

for sample in all_samples:
    mean_of_samples.append(np.mean(sample))
mean_of_samples
```

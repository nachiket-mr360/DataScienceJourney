What is Seaborn?
Seaborn is a high-level data visualization library built on top of Matplotlib.
It provides:

Beautiful, modern default styles
Concise syntax (less code for complex visuals)
Integration with Pandas DataFrames
Perfect for quick, attractive plots while keeping full Matplotlib power underneath.

What Are Relational Plots?
Relational plots are used to show the relationship between two variables.

They focus on how one variable changes with respect to another.
Typically used for continuous (numeric) data, but can also mix with categories.
In Seaborn, the main relational plots are:

scatterplot() → displays individual data points
lineplot() → shows a continuous trend or summary line
Both functions belong to Seaborn’s relational family.

```python
#we can load sample dataset from sns,
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset('iris')
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins')
```

```python
iris
```

```python
penguins
```

```python
#scatter plot with scatterplot()
sns.scatterplot(x=tips['total_bill'], y=tips['tip'])
#without providing label we can get it we can overide it and also wecan give title
plt.title('scatter plot showing tips') 
#we use plt in sns
```

```python
tips['total_bill'].corr(tips['tip'])
```

```python
#line plot with lineplot()
sns.lineplot(x=tips['day'],y= penguins['bill_depth_mm'])
#we can also compare different data present in different table or file
```

```python
sns.lineplot(x=tips['day'], y=tips['tip'], estimator='mean', markers='o') #gives average 
```

```python
#syling and colours palates
sns.set_style('darkgrid') #options, white, dark, ticks, etc
sns.set_palette('pastel')
sns.scatterplot(x=tips['total_bill'], y=tips['tip'], hue=tips['sex'])
#using hue we can differentiate between things based on give value.
```

Categorical Plots

What Are Categorical Plots?
Categorical plots visualize data where at least one variable is categorical
(e.g., day of week, gender, class).

Seaborn gives several easy options to:

Compare counts or averages across categories
Understand spread & distribution
barplot() vs countplot()

```python
tips
```

```python
sns.barplot(x='day', y='total_bill', data=tips) #we can write like this also
#bar plot give average on th basis of category
```

```python


sns.countplot(x=tips['day']) 
#ti only count the things 
```

```python
#like 
tips['day'].value_counts()
```

```python
#boxplot and violinplot()
sns.boxenplot(x='day', y='total_bill', data=tips)
```

```python
sns.violinplot(x='day', y='total_bill', data=tips)
```

```python
#stripplot and swarmplot
sns.stripplot(x='day',y='total_bill', data=tips, jitter=True) #to protect overlapping
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips) #to avoid overlap totally
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips) #to avoid overlap totally
sns.stripplot(x='day',y='total_bill', data=tips, jitter=True) #to protect overlapping
```

| Plot        | Best for |
|-------------|-----------|
| **barplot** | Compare average values across categories |
| **countplot** | Show how many items fall in each category |
| **boxplot** | See median, quartiles, and outliers |
| **violinplot** | Same as boxplot + shows distribution |
| **stripplot** | View every observation (good for small data) |
| **swarmplot** | Like stripplot but avoids overlap |




Distribution Plots

What Are Distribution Plots?
Distribution plots show how data values are spread or concentrated.

Seaborn offers:


histplot() → modern histogram (with optional KDE)
kdeplot() → smooth curve of distribution
rugplot() → small marks showing individual values
(Older distplot() is now replaced by histplot().)
histplot() – Basic Histogram

```python
tips.head()
```

```python
sns.histplot(tips['total_bill'])
```

```python
sns.histplot(tips['tip'])
```

```python
#we can set bins
sns.histplot(tips['total_bill'], bins=30, kde=True) #kde gives a line, we can know skewness and all things
plt.title('histogram of total ll')
```

```python
#kde karnal density estimator plot
sns.kdeplot(tips['total_bill'], fill=True, color='green')
```

```python
#distplot (legacy, nowweusehistplotorkdeplot instead of t)vs histplot()
sns.displot(tips['total_bill'])
```

```python
#rugplot (small ticks for observations)
sns.rugplot(tips['total_bill'], height=0.4, color='red')
```

```python
sns.histplot(x=tips['total_bill'], hue=tips['sex'], kde=True, bins=20)
```

Key takeaways

histplot() → flexible histogram (can combine with KDE).

kdeplot() → smooth density curve.

rugplot() → quick look at raw values.

distplot() is deprecated → prefer histplot().

Use hue to compare groups in the same distribution plot.

Regression & Mixed Plots

Regression & Mixed Plots
Seaborn provides plots that combine relationships and statistical models or mix multiple plot types.

regplot() / lmplot() → scatter + regression line

jointplot() → scatter with hist/KDE

pairplot() → multiple relationships in one go

regplot() – Scatter + Regression Line

```python
sns.regplot(x=tips['total_bill'],y=tips['tip'],color='green') #we get a line called regrassion line (can't use hue)
```

```python
#lmplort() reggression with facets
sns.lmplot(x='total_bill', y='tip', data=tips, height=5, aspect=1.2)#similar to regplot but have height aspect points and use hue but not in regplot
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips,hue='sex', height=5, aspect=1.2)
```

```python
#jointplot() - scatter +histogra/kde
sns.jointplot(x='total_bill', y='tip', data=tips)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg')
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='scatter')
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde')
```

```python
#pairplot - relationship between many variabl
sns.pairplot(iris)
```

```python
sns.pairplot(iris, hue='species')
```
